```python
import csv
import pandas as pd
from pathlib import Path

# Ruta del archivo CSV con los problemas
INPUT_FILENAME = r"C:\Users\mkoro\Desktop\Transformers\test_dataset\data\processed\leetcode_problems_processed_data - Copy.csv"

# Ruta del archivo Excel de salida
OUTPUT_EXCEL = r"C:\Users\mkoro\Desktop\Transformers\test_dataset\outputs\prompts_generados.xlsx"

# Número de prompts a generar
NUM_PROMPTS = 15

# Plantilla del prompt
def generar_prompt(description: str) -> str:
    return (
        "Resuelve el siguiente problema en Python. La solución debe implementar una función que, "
        "dado(s) input(s) especificado(s) en el problema, compare su output con el output esperado y "
        "ejecute dichos tests. La función debe imprimir 'True' por cada test que pase y 'False' por cada "
        "test que falle, y finalmente imprimir la cantidad de tests correctos sobre el total. "
        "Proporciona únicamente código Python ejecutable.\n\n"
        f"{description}"
    )

# Leer los problemas del archivo CSV
def leer_problemas_csv(ruta_csv: str, limite: int = 15):
    problemas = []
    with open(ruta_csv, mode='r', encoding='utf-8') as archivo:
        lector_csv = csv.reader(archivo)
        next(lector_csv, None)  # Salta la cabecera
        for i, campos in enumerate(lector_csv):
            if len(campos) < 2:
                continue
            if i >= limite:
                break
            problema = {
                "ID": campos[0],
                "Descripción": campos[1]
            }
            problemas.append(problema)
    return problemas

def main():
    problemas = leer_problemas_csv(INPUT_FILENAME, limite=NUM_PROMPTS)
    
    # Generar los prompts
    prompts = []
    for problema in problemas:
        prompt_texto = generar_prompt(problema["Descripción"])
        prompts.append({
            "ID": problema["ID"],
            "Prompt": prompt_texto
        })
    
    # Guardar los prompts en un archivo Excel
    df = pd.DataFrame(prompts)
    Path(OUTPUT_EXCEL).parent.mkdir(parents=True, exist_ok=True)
    df.to_excel(OUTPUT_EXCEL, index=False)
    print(f"Prompts guardados en el archivo: {OUTPUT_EXCEL}")

if __name__ == "__main__":
    main()
```
